```python
import networkx as nx
import time as time
import ast
import csv

def list2nx(edgelist):
    G = nx.Graph()
    G.add_edges_from((edgelist))
    return G
```

```python
graph_list = []
with open("data/bd_mers.csv", "r") as f:
    reader = csv.reader(f)

    for row in reader:
        if not row[0].isdigit():
            continue
        graph_list.append(row)


bd_mer_list = []
for _,_,edges in graph_list:

    g = list2nx(ast.literal_eval(edges))
    bd_mer_list.append(g)

print(len(bd_mer_list))
```
